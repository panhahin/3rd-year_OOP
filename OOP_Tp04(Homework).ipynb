- Create a class Book with title, author, and price attributes.
- Demonstrate encapsulation by protecting the price attribute.
- Use inheritance to create an EBook subclass.

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price      # private price

    # Encapsulation: we use this methods to read and change the private price
    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price >= 0:
            self.__price = new_price
        else:
            print("Price cannot be negative!")

    def show(self):
        print("Book:", self.title, "by", self.author, "- Price:", self.__price)


# child of Book
class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)   # reuse Book's code
        self.file_size = file_size               # new attribute for EBook

    def show(self):
        print("EBook:", self.title, "by", self.author, "- Price:", self.get_price(), "- Size:", self.file_size, "Mb")


# Test the classes
book1 = Book("Math", "Mr. Panha", 20)
ebook1 = EBook("Calculus", "Dr. Panha", 12, 3)

book1.show()
ebook1.show()

print(book1.get_price())    # 20
book1.set_price(25)         # change price
print(book1.get_price())    # 25
book1.set_price(-5)         # wrong price -> message
```
