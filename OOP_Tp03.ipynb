```python
class Greeting:
    """A class for greeting a user."""

    def __init__(self, name):
        self.name = name

    def say_hello(self):
        """Print a greeting message."""
        print(f"Hello, {self.name}!")


def main():
    """Get the user's name and display a greeting."""
    name = input("Enter your name: ").strip()

    greeting = Greeting(name)
    greeting.say_hello()


if __name__ == "__main__":
    main()
```

- **Exercise 2**

```python
class Rectangle:
    """A rectangle with a length and a width."""

    def __init__(self, length, width):
        if length <= 0 or width <= 0:
            raise ValueError("length and width must be greater than 0")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def display(self):
        print(f"Length    : {self.length}")
        print(f"Width     : {self.width}")
        print(f"Area      : {self.area()}")
        print(f"Perimeter : {self.perimeter()}")


def read_number(message):
    """Ask until the user types a valid number."""
    while True:
        try:
            return float(input(message))
        except ValueError:
            print("Please enter a number, for example 5 or 3.5")


def main():
    length = read_number("Enter the length of the rectangle: ")
    width = read_number("Enter the width of the rectangle: ")
    try:
        rect = Rectangle(length, width)
    except ValueError as error:
        print("Error:", error)
    else:
        rect.display()


if __name__ == "__main__":
    main()
```

```python
class EquationCalculator: #Creat class name "EquationCalculator"
    def __init__(self, y, z): #Constructor initializes the values of y and z
        self.y = y #store the input values
        self.z = z 

    def calculate_x(self): #Creat the method to 'calculate x" by using the formula
        if self.y == 0 or self.z == 0: #Using condition to identify y, z > 0
             raise ValueError("y and z must not be zero.") #Manually trigger exception(error)
        if self.y + self.z == 0: #Identify y + z > 0
            raise ValueError("y + z must not be zero")
        x = (self.y * self.z)/(self.y + self.z) #Method of equation
        return x #sent the calculate result back to the caller 

    def display_result(self): #Creat method to display x
        x = self.calculate_x() #Call function calculate_x
        print(f"x = {x}") #Display value of x

try: #Using try block lets you test a block of code errors
    #Get input from user 
    y = float(input("Enter a value of y: "))
    z = float(input("Enter a value of z: "))

     #Creat  calculator object 
    calculator = EquationCalculator(y, z)

    calculator.display_result() #Display the result
except ValueError: #Using except block lets you handle the error
    print("Error: Please enter the valid number!")
```

```python
class HundredsCounter:
    def __init__(self, number):
        if number < 0:
            raise ValueError("Number must be non-negative.")

        self.number = number

    def count_hundreds(self):
        return self.number // 100


# Create an object
counter = HundredsCounter(999)

# Display the result
print("Number:", counter.number)
print("Complete hundreds:", counter.count_hundreds())
```

```python
# Exercise 5: Cheater game (display the next number)
class LuckyNumberGame:
    def __init__(self, number):
        # store the number provided by the user
        self.number = number

    def get_next_number(self):
        # Return the integer immediately after the stored number.
        return self.number + 1


def ask_to_play_again():
    # Keep asking until the user enters a valid yes-or-no answer.
    while True:
        answer = input("Do you want to play again? (Yes/No): ").strip().lower()

        if answer in {"yes", "no"}:
            return answer == "yes"
        
        print("Please enter yes or no.")

def main():
    # Run another round until the user chooses to stop .
    while True:
        try:
            # Convert the user's input to an integer for the game.
            number = int(input("Please enter an integer: "))

        except ValueError:
            # A non-integer input is rejected; the loop will prompt again.
            print("Please enter a valid integer.")

        else:
            # Create the game and display the next integer.
            game = LuckyNumberGame(number)
            print(f"The next number is: {game.get_next_number()}")
            
        # ask the user if they want to play again but if they say no, break the loop and end the game
        if not ask_to_play_again():
            # stop loop
            break

# Start the game only when this file is run directly.
if __name__ == "__main__":
    main()
```
